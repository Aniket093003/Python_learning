```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
sales_data = np.array([
    [1,1500,1800,2000,2200],
    [2,1200,1400,1600,1800],
    [3,2000,2300,2600,2900],
    [4,1800,2100,2400,2700],
    [5,1600,1850,2050,2300]
])

print("=== Sales analysis ===")
print("\n sales data shape",sales_data.shape)
print("\n sales data of first 3 restraurent \n",sales_data[:3, : ])
```

```python
### total sales per year

# print(np.sum(sales_data , axis=0))
yearly_total = np.sum(sales_data[:,1:], axis=0)
print(yearly_total)
min_sales = np.min(sales_data[:,1:],axis=1)
max_sales = np.max(sales_data[:,1:],axis=0)
print("min sales",min_sales)
print("max sales per year",max_sales)

avg_sales = np.mean(sales_data[: , 1:],axis=1)
print("avg sales per restraunt", avg_sales)

monthly_sales = sales_data[:,1:] / 12
print("monthly sales \n",monthly_sales)
```
